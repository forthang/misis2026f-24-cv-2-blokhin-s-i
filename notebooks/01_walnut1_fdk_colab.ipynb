## 1. Проверить GPU
Нужна NVIDIA GPU.  ПРОВВЕРЬ ТИП СРЕДЫ 

```python
import shutil
import subprocess
import sys

if shutil.which('nvidia-smi') is None:
    raise RuntimeError('GPU не подключена. Выбери Runtime → Change runtime type → GPU и повтори запуск.')
gpu_info = subprocess.check_output(
    ['nvidia-smi', '--query-gpu=name,memory.total,driver_version', '--format=csv,noheader'],
    text=True,
).strip()
print('Python:', sys.version.split()[0])
print('GPU, память, драйвер:', gpu_info)
```

## 2. Установить библиотеки
ASTRA выполняет 3D-реконструкцию на GPU. Остальные библиотеки читают TIFF и показывают результат.

```python
subprocess.check_call([
    sys.executable, '-m', 'pip', 'install', '-q',
    'astra-toolbox==2.5.0', 'imageio==2.38.0',
    'matplotlib>=3.8,<4', 'tifffile', 'ipywidgets>=8,<9',
])
import astra
if not astra.use_cuda():
    raise RuntimeError('ASTRA не видит CUDA. Проверь, что выбрана GPU-среда.')
print('ASTRA:', astra.__version__)
print(astra.get_gpu_info())
```
